```python
#all 1s 
np.ones((1,3,3))
```

```python
np.full((2,2),69)
```

```python
np.full_like(a,7)
```

```python
#random decimal number
np.random.rand(2,3,1)
```

```python
#for Int value
np.random.randint(-4,5,size=(3,3))
```

```python
#Identity matrix
np.identity(3)
```

```python
arr = np.array([[1,2,3]])
r1 = np.repeat(arr,3,axis=0)
print(r1)
```

```python
new = np.ones((5,5))
print(new)
x = np.zeros((3,3))
x[1,1]=9
print(x)
new[1:-1,1:-1]=x
print(new)
```

##### Recognizing Arrays

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)

after = before.reshape((4,2))
print(after)
```

```python
# vertically stacking vectors

v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])
np.vstack([v1,v2,v1,v2])
```

```python
h1 = np.ones((2,4))
h2 = np.zeros((2,2))
np.hstack((h1,h2))
```
